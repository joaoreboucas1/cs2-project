```python
import numpy as np
import matplotlib.pyplot as plt
import utils
```

```python
minima = {index: utils.read_minimum(index) for index in [48, 94, 95, 96, 97]}
aktypes = [0, 1, 2, 4, 3]
for (index, minimum), aktype in zip(minima.items(), aktypes):
    minimum['aktype'] = aktype
```

```python
for index, minimum in minima.items():
    print(f"{index} => {utils.get_chi2(minimum)}")
```

```python
data = {}
for index, minimum in minima.items():
    data[index] = utils.get_cmb(minimum)
```

```python
cl_tt_gr = data[48][4]
for index, spectra in data.items():
    ells = np.arange(len(spectra[4])) + 2
    plt.plot(ells, (spectra[4]-cl_tt_gr)/cl_tt_gr, color='black' if index == 48 else utils.colors[(index-96)%4])
plt.ylim([-0.05, 0.05])
```
